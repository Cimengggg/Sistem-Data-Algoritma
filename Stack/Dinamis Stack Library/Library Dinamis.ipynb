Library Stack Dinamis

```python
# Library Stack Dinamis Menggunakan List Python

class StackBukuDinamis:
    def __init__(self):
        # Menggunakan list kosong, secara default ukurannya dinamis
        self.data = []

    def kosong(self):
        return len(self.data) == 0

    def tambah_buku(self, nama_buku):
        # append() otomatis menambahkan data ke ujung (puncak stack) tanpa batas kapasitas
        self.data.append(nama_buku)
        print("Buku berhasil ditambahkan.")

    def ambil_buku(self):
        if self.kosong():
            print("Tumpukan buku kosong!")
        else:
            # pop() mengambil dan menghapus elemen terakhir dari list
            buku = self.data.pop()
            print("Buku yang diambil:", buku)

    def tampilkan_buku(self):
        if self.kosong():
            print("Tumpukan buku kosong!")
        else:
            print("Tumpukan buku dari atas:")
            # Menampilkan dari indeks terakhir (puncak) ke indeks pertama (dasar)
            for i in range(len(self.data) - 1, -1, -1):
                print("-", self.data[i])
```
